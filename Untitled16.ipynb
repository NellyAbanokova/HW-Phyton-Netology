```python
# ДЗ 4
books = [
    {'genre': 'поэзия', 'number': '978-5-1000-1234-7', 'title': 'Евгений Онегин', 'author': 'Александр Пушкин'},
    {'genre': 'фэнтези', 'number': '88006', 'title': 'Властелин колец', 'author': 'Джон Р. Р. Толкин'},
    {'genre': 'детектив', 'number': 'D-1122', 'title': 'Безмолвный свидетель', 'author': 'Агата Кристи'}
]

directories = {
    '1': ['978-5-1000-1234-7', '88006'],
    '2': ['D-1122'],
    '3': []
}
```

```python
def find_book_by_number(number):
    for book in books:
        if book['number'] == number:
            return book
    return None


def find_book_by_title(title):
    for book in books:
        if book['title'] == title:
            return book
    return None


def find_shelf_by_number(number):
    for shelf, nums in directories.items():
        if number in nums:
            return shelf
    return None


def get_all_shelves():
    return ', '.join(directories.keys())
```

```python
def book_info():
    number = input('Введите номер книги:\n')
    book = find_book_by_number(number)

    if book:
        print(f"Название книги: {book['title']}")
        print(f"Автор: {book['author']}")
    else:
        print('Книга не найдена в базе')
```

```python
def shelf():
    title = input('Введите название книги:\n')
    book = find_book_by_title(title)

    if not book:
        print('Книга не найдена в базе')
        return

    shelf_number = find_shelf_by_number(book['number'])

    if shelf_number:
        print(f'Книга хранится на полке: {shelf_number}')
    else:
        print('Книга не найдена в базе')
```

```python
def show_all():
    for book in books:
        shelf_number = find_shelf_by_number(book['number'])
        print(
            f"№: {book['number']}, жанр: {book['genre']}, "
            f"название: {book['title']}, автор: {book['author']}, "
            f"полка хранения: {shelf_number}"
        )
```

```python
def add_shelf():
    shelf = input('Введите номер полки:\n')

    if shelf in directories:
        print(f"Такая полка уже существует. Текущий перечень полок: {get_all_shelves()}.")
    else:
        directories.setdefault(shelf, [])
        print(f"Полка добавлена. Текущий перечень полок: {get_all_shelves()}.")
```

```python
while True:
    command = input('Введите команду:\n')

    match command:
        case 'book_info':
            book_info()

        case 'shelf':
            shelf()

        case 'all':
            show_all()

        case 'add_shelf':
            add_shelf()

        case 'del_shelf':
            del_shelf()

        case 'q':
            print('Программа завершена')
            break

        case _:
            print('Неизвестная команда')
```
